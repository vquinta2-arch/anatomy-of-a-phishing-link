```r
library(dplyr)
library(tidyr)
library(tibble)

# 1. The 56 URL-structure features
url_features <- df %>% select(length_url:statistical_report) %>% names()
length(url_features)   # should be 56

# 2. Theme map (my assignment from the column names; edit freely)
themes <- list(
  "1. Hiding the destination" = c(
    "nb_at", "ip", "punycode", "port", "shortening_service", "nb_dslash",
    "nb_redirection", "nb_external_redirection", "nb_percent"),
  "2. Borrowing trust" = c(
    "brand_in_subdomain", "brand_in_path", "domain_in_brand", "tld_in_subdomain",
    "tld_in_path", "https_token", "http_in_path", "nb_com", "nb_www", "phish_hints",
    "suspecious_tld", "prefix_suffix", "abnormal_subdomain", "random_domain"),
  "3. Query string and special characters" = c(
    "nb_qm", "nb_eq", "nb_and", "nb_or", "nb_semicolumn", "nb_dollar", "nb_star",
    "nb_space", "nb_tilde", "nb_comma", "nb_underscore"),
  "4. Padding and clutter" = c(
    "length_url", "length_hostname", "nb_slash", "nb_dots", "nb_subdomains",
    "nb_hyphens", "nb_colon", "ratio_digits_url", "ratio_digits_host", "char_repeat",
    "length_words_raw", "shortest_words_raw", "shortest_word_host", "shortest_word_path",
    "longest_words_raw", "longest_word_host", "longest_word_path",
    "avg_words_raw", "avg_word_host", "avg_word_path"),
  "5. Other" = c("path_extension", "statistical_report")
)

theme_map <- tibble(
  theme       = rep(names(themes), lengths(themes)),
  theme_order = rep(seq_along(themes), lengths(themes)),
  feature     = unlist(themes, use.names = FALSE)
)

# Every feature must have exactly one theme
missing_feats <- setdiff(url_features, theme_map$feature)
extra_feats   <- setdiff(theme_map$feature, url_features)
print(missing_feats); print(extra_feats)   # both should be character(0)
stopifnot(length(missing_feats) == 0, length(extra_feats) == 0,
          !anyDuplicated(theme_map$feature))

# 3. Per feature and class: usage and mean
stats <- df %>%
  select(status, all_of(url_features)) %>%
  pivot_longer(-status, names_to = "feature", values_to = "value") %>%
  group_by(feature, status) %>%
  summarise(n_sites    = n(),
            n_uses     = sum(value > 0, na.rm = TRUE),
            mean_value = round(mean(value, na.rm = TRUE), 3),
            .groups = "drop") %>%
  mutate(pct_uses     = round(n_uses / n_sites * 100, 2),
         pct_not_uses = round(100 - pct_uses, 2))

# 4. Per feature: gap, ratios, theme, rankings
feat <- stats %>%
  select(feature, status, pct_uses, mean_value) %>%
  pivot_wider(names_from = status, values_from = c(pct_uses, mean_value)) %>%
  left_join(theme_map, by = "feature") %>%
  mutate(gap_pp   = round(pct_uses_phishing - pct_uses_legitimate, 2),  # + = phishing, - = legitimate
         abs_gap  = abs(gap_pp),
         leans_to = case_when(gap_pp > 0 ~ "Phishing",
                              gap_pp < 0 ~ "Legitimate",
                              TRUE       ~ "Neither"),
         usage_type  = ifelse(pct_uses_legitimate >= 95 & pct_uses_phishing >= 95,
                              "Always present", "Optional"),
         usage_ratio = round((pct_uses_phishing + 0.01) / (pct_uses_legitimate + 0.01), 2),
         mean_ratio  = round((mean_value_phishing + 0.001) / (mean_value_legitimate + 0.001), 2),
         feature_label = ifelse(feature %in% c("longest_word_path", "avg_word_path",
                                               "shortest_word_path"),
                                paste0(feature, " (= has a path)"), feature)) %>%
  # by theme, then biggest usage gap; mean_ratio only breaks ties
  # (always-present features all have a usage gap of 0)
  arrange(theme_order, desc(abs_gap), desc(abs(log(mean_ratio)))) %>%
  group_by(theme) %>%
  mutate(rank_in_theme = row_number()) %>%
  ungroup() %>%
  mutate(feature_order = row_number())   # one sort key for the whole table

# Overall gap rank for the Top 5 / 10 / 15 charts (Optional features only;
# two of the three path-word features are duplicates and are excluded)
eligible <- feat$usage_type == "Optional" &
  !feat$feature %in% c("avg_word_path", "shortest_word_path")
feat$rank_gap <- NA_integer_
feat$rank_gap[eligible] <- rank(-feat$abs_gap[eligible], ties.method = "first")

# 5. One Tableau-ready table: one row per feature x status (56 x 2 = 112 rows)
final <- stats %>%
  left_join(feat %>% select(feature, feature_label, theme, theme_order, rank_in_theme,
                            feature_order, usage_type, gap_pp, abs_gap, leans_to,
                            usage_ratio, mean_ratio, rank_gap),
            by = "feature") %>%
  mutate(status = ifelse(status == "phishing", "Phishing", "Legitimate")) %>%
  arrange(feature_order, status) %>%
  select(theme, theme_order, rank_in_theme, feature_order, feature, feature_label,
         usage_type, status, pct_uses, pct_not_uses, n_uses, n_sites, mean_value,
         gap_pp, abs_gap, leans_to, usage_ratio, mean_ratio, rank_gap)

# 6. Checks, then save
stopifnot(nrow(final) == 112,
          all(final$pct_uses >= 0 & final$pct_uses <= 100),
          all(abs(final$pct_uses + final$pct_not_uses - 100) < 0.01))

write.csv(final, "phishig_datasets_syntax_analysis.csv", row.names = FALSE)
getwd()   # folder where the file was saved

options(repr.matrix.max.rows = 120, repr.matrix.max.cols = 20, width = 200)
final
```
